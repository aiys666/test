## Step 1：导入模块

```python
import sys
import os
import pandas as pd

sys.path.insert(0, r'E:\岳松个人文件\嘉银工作内容的整理\code')
from mexico_presto import PrestoQuery

print('✅ 模块导入成功')
```

---
## Step 2：初始化 & 登录

```python
pq = PrestoQuery()
pq.login()
print('✅ 登录成功')
```

---
## Step 3：编写 SQL

> 改下面这行，想查什么写什么。

```python
SQL = """
select * from mx.gm_ltv_shou_user_media_source
""".strip()

print(f'📋 SQL:\n{SQL}')
```

---
## Step 4：执行查询

```python
result = pq.query(SQL)

if isinstance(result, pd.DataFrame):
    df = result
    print(f'✅ 查询完成  →  {df.shape[0]:,} 行 × {df.shape[1]} 列')
    print(f'📊 内存占用: {df.memory_usage(deep=True).sum() / 1024**2:.1f} MB')
else:
    print(f'❌ 查询失败: {result}')
    df = None
```

---
## Step 5：预览数据

```python
if df is not None and not df.empty:
    print('📋 列名:')
    print(df.columns.tolist())
    print(f'\n🔍 前5行:')
    display(df.head())
    print(f'\n📊 数据类型:')
    print(df.dtypes.to_string())
else:
    print('⚠️ 无数据可预览')
```

---
## Step 6：导出 Excel

```python
if df is not None and not df.empty:
    ts = pd.Timestamp.now().strftime('%Y%m%d_%H%M%S')
    out = f'墨西哥_{ts}.xlsx'
    df.to_excel(out, index=False)
    print(f'💾 已保存  →  {os.path.abspath(out)}')
    print(f'📏 {len(df):,} 行 × {len(df.columns)} 列')
else:
    print('⚠️ 无数据可导出')
```
